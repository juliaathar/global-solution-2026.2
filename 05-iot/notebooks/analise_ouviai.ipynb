# OuviAí: análise dos dados e dos resultados

Roda os scripts de `../codigo` e mostra os resultados. Preencha antes `evidencias/model_testing.csv` e `evidencias/tentativas_ao_vivo.csv`.

```
import subprocess, sys
from pathlib import Path
from IPython.display import Markdown, Image, display
COD = Path('..') / 'codigo'
EVI = Path('..') / 'evidencias'
def rodar(*args):
    r = subprocess.run([sys.executable, *args], cwd=COD, capture_output=True, text=True)
    print(r.stdout[-3000:] or r.stderr[-3000:])
```

## 1. Dataset: formato, contagem e reuso

```
rodar('verificar_dataset.py')
```

## 2. Model testing (15 arquivos de Test)

```
rodar('avaliar.py', 'modelo')
if (EVI / 'resultado_model_testing_matriz.png').exists():
    display(Image(EVI / 'resultado_model_testing_matriz.png'))
```

## 3. Teste ao vivo (15 tentativas)

```
rodar('avaliar.py', 'ao-vivo')
if (EVI / 'resultado_ao_vivo.md').exists():
    display(Markdown((EVI / 'resultado_ao_vivo.md').read_text(encoding='utf-8')))
```
